```python
para = """Artificial intelligence is transforming the way we interact with technology. It enables machines to learn from data and improve over time. Natural language processing helps computers understand human language. This technology is widely used in chatbots and virtual assistants. Many industries are adopting AI to increase efficiency and productivity. In healthcare, AI assists doctors in diagnosing diseases more accurately. In finance, it helps detect fraud and manage risks effectively. Education is also benefiting from personalized learning systems powered by AI. However, ethical concerns such as data privacy and bias must be addressed. Overall, AI continues to shape the future of innovation and human progress."""
```

```python
import nltk

nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords

stemmer = PorterStemmer()
sentences = sent_tokenize(para)
stop_words = set(stopwords.words("english"))
```

```python
## apply stopwwords
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    words = [word for word in words if word.lower() not in stop_words]
    pos_tags = nltk.pos_tag(words)
    print(pos_tags)

    
```

Named Entity Recognition (NER)

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
import nltk
from nltk.tokenize import word_tokenize

sentence = "Taj Mahal is in Agra and was built by Shah Jahan"

words = word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

named_entities = nltk.ne_chunk(pos_tags)

print(named_entities)
```
