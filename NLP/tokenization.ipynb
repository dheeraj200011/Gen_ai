Porter Stemmer

```python
## stemming - ye basically ccommon word ko catch karta hai
## 1> porter stemming

from nltk.stem import PorterStemmer

stemming = PorterStemmer()

words = ["playing", "played", "plays", "runner", "running"]

for word in words:
    print(word, "=>", stemming.stem(word))
 
```

Regexp Stemmer

```python
## isme hum regex ki help se suffix ko get karte hai
## isme hum word ki length ko bhi justifykarte hai

from nltk.stem import RegexpStemmer
stemmer = RegexpStemmer('ing$|s$|ed$', min = 4)

words = ["playing", "played", "plays", "cats", "dog"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

Snowball Stemmer

```python
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

words = ["playing", "played", "plays", "running", "happiness"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
## Apply stopwords

import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

# setup
stemmer = PorterStemmer()
stop_words = set(stopwords.words("english"))

# example sentences list
sentences = [
    "This is a good book",
    "I am learning natural language processing"
]

# processing
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])
    
    words = [stemmer.stem(word) for word in words if word.lower() not in stop_words]
    
    sentences[i] = " ".join(words)

print(sentences)

```

```python
from nltk.stem import SnowballStemmer
snowballstemmer=SnowballStemmer('english')
```

```python
## Apply Stopwords And Filter And then Apply Snowball Stemming

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[snowballstemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)# converting all the list of words into sentences
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
## Apply Stopwords And Filter And then Apply Snowball Stemming

for i in range(len(sentences)):
    #sentences[i]=sentences[i].lower()
    words=nltk.word_tokenize(sentences[i])
    words=[lemmatizer.lemmatize(word.lower(),pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)# converting all the list of words into sentences
```
