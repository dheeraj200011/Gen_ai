Bag of Words

Ye ek technique hai jo text (sentence/document) ko numbers (vectors) mein convert karti hai.

👉 Isme hum:

Words ka order ignore kar dete hain ❌
Sirf kaunse words aaye aur kitni baar aaye (frequency) dekhte hain ✅

"I love coding"
"I love AI"
[I, love, coding, AI]

```python
## bag of words implementation

import pandas as pd 

message = pd.read_csv("spam.csv", encoding='latin-1')
message = message[['v1', 'v2']]
message.columns = ['label', 'message']

print(message)
```

```python
## data cleaning and preprocessing

import re
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer

stemmer = SnowballStemmer("english")

stop_words = set(stopwords.words("english"))

corpus = []

for i in range(0, len(message)):
    review = re.sub('[^a-zA-Z]', " ", message["message"][i])
    review = review.lower()
    review = review.split()
    review = [stemmer.stem(word) for word in review if not word not in stop_words]
    review = " ".join(review)
    corpus.append(review)
```

```python
corpus
```

```python
## create the bag of words

from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=100)

x= cv.fit_transform(corpus)

# words
print("Vocabulary:", cv.get_feature_names_out())

# matrix
print("BoW Matrix:\n", x.toarray())
```

N GRAMS

👉 N-Grams ek NLP technique hai jisme hum continuous (lagataar) words ke group banate hain.
👉 “N” ka matlab = kitne words ek saath liye ja rahe hain
BASICALLY YE PAIRRS BNATA HAI

Sentence:

"I love AI and coding"

👉 Unigram:

[I, love, AI, and, coding]

👉 Bigram:

[I love, love AI, AI and, and coding]

👉 Trigram:

[I love AI, love AI and, AI and coding]

```python
## n gram implementation

from sklearn.feature_extraction.text import CountVectorizer
sentences = ["I love coding", "I love AI"]
cv = CountVectorizer(ngram_range=(1,2))
x = cv.fit_transform(sentences)

print(cv.get_feature_names_out())
print(x.toarray())
```

TF = Term Frequency
IDF = Inverse Document Frequency

isme hum sentences me jo unique hai wo important hota hai or jo repeat poar ho wo less important hota hai

```python
## implemt tf/idf
## best practice hai tf/idf ke sath n_grams ko use karna

from sklearn.feature_extraction.text import TfidfVectorizer

tdidf = TfidfVectorizer(max_features=200)
X = tdidf.fit_transform(corpus)

print(tdidf.get_feature_names_out())
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tdidf = TfidfVectorizer(max_features=100, ngram_range=(1,2))
X = tdidf.fit_transform(corpus)

print(tdidf.get_feature_names_out())
print(X.toarray())
```
