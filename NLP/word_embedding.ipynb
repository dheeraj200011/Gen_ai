Word Embedding

👉 Word Embedding ek advanced technique hai jisme words ko numbers (vectors) me convert kiya jata hai
👉 Lekin normal encoding (BoW/TF-IDF) ki tarah nahi — yeh meaning (semantics) bhi samajhta hai 🧠

CBOW (Continuous Bag of Words)

CBOW Word2Vec ka ek model hai jo surrounding words (context) ko use karke missing/current word predict karta hai.

Simple language me:

"Aas paas ke words dekh kar beech wala word guess karna"

Real Life Intuition

Sentence:

"I love playing ____ in the evening"

Aap easily guess kar sakte ho ki blank me "cricket" ya "football" ho sakta hai.

CBOW bhi exactly yehi karta hai.

Context words → Target word predict.

```python
## word2vec implementation

from gensim.models import Word2Vec

sentences = [
    ["i", "love", "nlp"],
    ["nlp", "is", "fun"],
    ["i", "love", "machine", "learning"]
]

model = Word2Vec(sentences, vector_size=100, window=2, min_count=1, sg=1)

print(model.wv["nlp"])
print(model.wv.most_similar("nlp"))
```
