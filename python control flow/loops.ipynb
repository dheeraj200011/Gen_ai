while loop

```python
## while loop

## The while loop continues to execute as long as the condition is True.

count = 0

while(count < 5):
    print(count)
    count += 1
```

```python
## loop control statements
## break, continue, pass

for i in range(5):
    if(i == 3):
        break
    print(i)

```

```python

## Examples- Calculate the sum of first N natural numbers using a while and for loop

## while loop

n = 10
sum = 0
count = 1

while(count <= n):
    sum = count + sum
    count = count + 1

print(sum)
```

```python
## Examples- Calculate the sum of first N natural numbers using a while and for loop

## for loop

n=10   
sum=0

for i in range(11):
    sum = sum  + i

print(sum)
```

```python
## continue

## The continue statement skips the current iteration and continues with the next.
## ye odd number show karega

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass
## The pass statement is a null operation; it does nothing.

for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
## Nested loopss
## a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Example- Prime numbers between 1 and 100
## case num 1 se bada or num % i == 0 nhi hona chiaye
## prime number wo hote hai jo sirf do number se hi divide ho

for num in range(1,101):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```

Conclusion:

Loops are powerful constructs in Python that allow you to execute a block of code multiple times. By understanding and using for and while loops, along with loop control statements like break, continue, and pass, you can handle a wide range of programming tasks efficiently.
