Importing Modules in Python: Modules and Packages

In Python, modules and packages help organize and reuse code. Here's a comprehensive guide on how to import them.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)

```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
from math import *
print(sqrt(16))
print(pi)
```

```python
## packges me hum manual function create karte hai ye files ko optoimized rakta hai

from packages.maths import addition
addition (2,5)
```

```python
from packages import maths
maths.addition(2,3)
```

```python
## hum package ke andar aur bhi packages bna sakte hai

from packages.subpackages.mult import multiplication
multiplication(2, 4)
```

Conclusion

Importing modules and packages in Python allows you to organize your code, reuse functionalities, and keep your projects clean and manageable. By understanding how to import modules, specific functions, and use relative imports within packages, you can structure your Python applications more effectively.
