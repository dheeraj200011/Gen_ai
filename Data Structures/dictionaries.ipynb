Introduction to Dictionaries,

Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type.

```python
## empty dictionary

empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict()
```

```python
student = {
    "name": "Dheeraj",
    "age": 31
}

print(student["name"])
```

```python
# Single key is slways used

student={"name":"Krish","age":32,"name":24} ## yeb last wale ko priority dega
print(student)
```

```python
## accessing Dictionary Elements
student={"name":"Krish","age":32,"grade":'A'}
print(student)

## Accessing Dictionary elements
print(student['grade'])
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not Available"))
```

```python
## add new key:value

student["address"] = "India"
print(student)

# edit value

student["age"] = 34
print(student)
```

```python
## delete key and value pair

del student["address"]
```

```python
print(student)
```

```python

## Dictionary methods

keys = student.keys() ## get all keys
print(keys)
values = student.values() ## get all values
print(values)
items = student.items() ## get both above
print(items)
```

```python
## shallow copy
## isme hum jo bhi chnge karenge wo main me change nhi hoga

student_copy = student

student_copy["name" ] = "Dheeraj"
print(student_copy)
print(student)

## dono me change hua


```

```python
## implement shallow copy

student_copy1 = student.copy()
print(student_copy1)

student["name" ] = "Krish"

## this is shallow copy
print(student)
print(student_copy1)
```

```python
### Iterating Over Dictionaries
## You can use loops to iterate over dictionatries, keys,values,or items

for key in student.keys():
    print(key)
    
for values in student.values():
    print(values)
    

for key,value in student.items():
    print(key, value)
```

```python
## Nested Disctionaries
students={
    "student1":{"name":"Krish","age":32},
    "student2":{"name":"Peter","age":35}
}
print(students)
```

```python
## Access nested dictionaries elementss
print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
## Iterating over nested dictionaries

for student, details in students.items():
    print(f"{student}:{details}")
    for key, value in details.items():
        print(f"{key}: {value}")
```

```python
## Dictionary Comphrehension

squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
## Dictionary Comphrehension
## only even numbers

evens = {x:x**2 for x in range(5) if x % 2 == 0}
print(evens)
```

```python
## Merge 2 dictionaries into one

dict1={"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```

```python
## Practical Examples

## USe a dictionary to count he frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1

print(frequency)

```

Conclusion

Dictionaries are powerful tools in Python for managing key-value pairs. They are used in a variety of real-world scenarios, such as counting word frequency, grouping data, storing configuration settings, managing phonebooks, tracking inventory, and caching results. Understanding how to leverage dictionaries effectively can greatly enhance the efficiency and readability of your code.
